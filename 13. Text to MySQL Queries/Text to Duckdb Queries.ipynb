## Text to DuckDB Queries

- Building Q&A systems of SQL databases requires executing model-generated SQL queries
- Make sure your database have only Read-Only access to prevent any data loss
- We will use a local DuckDB database with employee data to demonstrate the queries
- DuckDB is an embedded analytical database that's perfect for local development and testing

````python
from dotenv import load_dotenv
import os

os.environ['KMP_DUPLICATE_LIB_OK'] = 'True'

load_dotenv()
````

````python
from langchain_ollama import ChatOllama 
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough 
from langchain_core.prompts import ChatPromptTemplate

from langchain_core.tools import tool


llm = ChatOllama(model='llama3.2:3b', base_url='http://localhost:11434')
llm.invoke('hi')
````

````python
# !pip install duckdb duckdb-engine

import duckdb
from sqlalchemy import create_engine, event
from langchain_community.utilities import SQLDatabase

# Create a shared in-memory DuckDB connection
# We'll create a file-based database but delete it after use
db_path = "employees_temp.duckdb"

# Create SQLAlchemy engine for DuckDB  
engine = create_engine(f"duckdb:///{db_path}")

# Use SQLDatabase with SQLAlchemy engine
db = SQLDatabase(engine=engine)

# Create a connection through the engine for our direct operations
with engine.connect() as connection:
    # Store the raw connection for later use
    raw_conn = connection.connection.dbapi_connection

print("DuckDB setup completed successfully!")
````

````python
# Create employees table
create_table_query = """
CREATE TABLE IF NOT EXISTS employees (
    emp_no INTEGER PRIMARY KEY,
    birth_date DATE,
    first_name VARCHAR(50),
    last_name VARCHAR(50),
    gender VARCHAR(1),
    hire_date DATE
);
"""

# Use the LangChain database connection to create the table
db.run(create_table_query)
print("Employees table created successfully!")
````

````python
# Generate and insert 100 rows of test employee data
import random
from datetime import date, timedelta

# Sample data for generating realistic employee records
first_names = [
    'John', 'Jane', 'Michael', 'Sarah', 'David', 'Emily', 'Robert', 'Lisa', 'James', 'Jessica',
    'William', 'Ashley', 'Christopher', 'Amanda', 'Matthew', 'Melissa', 'Daniel', 'Jennifer', 'Anthony', 'Michelle',
    'Mark', 'Kimberly', 'Donald', 'Amy', 'Steven', 'Angela', 'Kenneth', 'Brenda', 'Paul', 'Emma',
    'Joshua', 'Olivia', 'Kevin', 'Cynthia', 'Brian', 'Marie', 'George', 'Janet', 'Edward', 'Catherine',
    'Ronald', 'Frances', 'Timothy', 'Christine', 'Jason', 'Samantha', 'Jeffrey', 'Deborah', 'Ryan', 'Rachel'
]

last_names = [
    'Smith', 'Johnson', 'Williams', 'Brown', 'Jones', 'Garcia', 'Miller', 'Davis', 'Rodriguez', 'Martinez',
    'Hernandez', 'Lopez', 'Gonzalez', 'Wilson', 'Anderson', 'Thomas', 'Taylor', 'Moore', 'Jackson', 'Martin',
    'Lee', 'Perez', 'Thompson', 'White', 'Harris', 'Sanchez', 'Clark', 'Ramirez', 'Lewis', 'Robinson',
    'Walker', 'Young', 'Allen', 'King', 'Wright', 'Scott', 'Torres', 'Nguyen', 'Hill', 'Flores',
    'Green', 'Adams', 'Nelson', 'Baker', 'Hall', 'Rivera', 'Campbell', 'Mitchell', 'Carter', 'Roberts'
]

# Clear existing data
db.run("DELETE FROM employees")

# Generate insert values as a single query for better performance
insert_values = []
for i in range(1, 101):
    emp_no = 10000 + i
    
    # Generate random birth date (age between 25-65)
    years_ago = random.randint(25, 65)
    birth_date = date.today() - timedelta(days=years_ago * 365 + random.randint(0, 365))
    
    # Random names
    first_name = random.choice(first_names)
    last_name = random.choice(last_names)
    
    # Random gender
    gender = random.choice(['M', 'F'])
    
    # Generate hire date (between 1-30 years ago, but after birth date + 18 years)
    min_hire_age = 18
    max_hire_years_ago = min(30, years_ago - min_hire_age)
    hire_years_ago = random.randint(1, max_hire_years_ago) if max_hire_years_ago > 0 else 1
    hire_date = date.today() - timedelta(days=hire_years_ago * 365 + random.randint(0, 365))
    
    # Format for SQL insert
    insert_values.append(f"({emp_no}, '{birth_date}', '{first_name}', '{last_name}', '{gender}', '{hire_date}')")

# Create a bulk insert statement
bulk_insert = f"""
INSERT INTO employees (emp_no, birth_date, first_name, last_name, gender, hire_date)
VALUES {', '.join(insert_values[:10])}
"""

# Insert first 10 records
db.run(bulk_insert)

# Insert remaining records in batches of 10
for i in range(10, 100, 10):
    batch_values = insert_values[i:i+10]
    if batch_values:
        batch_insert = f"""
        INSERT INTO employees (emp_no, birth_date, first_name, last_name, gender, hire_date)
        VALUES {', '.join(batch_values)}
        """
        db.run(batch_insert)

print("Successfully inserted 100 employee records!")

# Verify the data
result = db.run("SELECT COUNT(*) FROM employees")
print(f"Total employees in database: {result}")

# Show first 5 records
print("\nFirst 5 employee records:")
first_five = db.run("SELECT * FROM employees ORDER BY emp_no LIMIT 5")
print(first_five)
````

### SQL Query Chains

- convert the question into a SQL query;
- execute the query;
- use the result to answer the original question.

````python
from langchain.chains import create_sql_query_chain

sql_chain = create_sql_query_chain(llm, db)
````

````python
sql_chain.get_prompts()[0].pretty_print()
````

````python
sql_chain
````

````python
question = "how many employees are there? You MUST RETURN ONLY DUCKDB QUERIES."
response = sql_chain.invoke({'question': question})
print(response)
````

### Extract Correct Query from LLM

````python
from scripts.llm import ask_llm
from langchain_core.runnables import chain
````

````python
@chain
def get_correct_sql_query(input):
    context = input['context']
    question = input['question']

    intruction = """
        Use above context to fetch the correct SQL query for following question
        {}

        Do not enclose query in ```sql and do not write preamble and explanation.
        You MUST return only single DuckDB SQL query.
    """.format(question)

    response = ask_llm(context=context, question=intruction)

    return response
````

````python
response = get_correct_sql_query.invoke({'context': response, 'question': question})
````

````python
db.run(response)
````

### Final SQL Query Chain 

````python
from langchain_community.tools.sql_database.tool import QuerySQLDataBaseTool
````

````python
execute_query = QuerySQLDataBaseTool(db=db)
sql_query = create_sql_query_chain(llm, db)

final_chain = (
    {'context': sql_query, 'question': RunnablePassthrough()}
    | get_correct_sql_query
    | execute_query
)
````

````python
question = "how many employees are there? You MUST RETURN ONLY DUCKDB QUERIES."

response = final_chain.invoke({'question': question})
print(response)
````

### MySQL Query Execution with Agents with LangGraph

- It can answer questions based on the databases' schema as well as on the databases' content (like describing a specific table).
- It can recover from errors by running a generated query, catching the traceback and regenerating it correctly.
- It can query the database as many times as needed to answer the user question.
- It will save tokens by only retrieving the schema from relevant tables.


````python
# %pip install --upgrade --quiet langgraph
````

````python
from langchain_community.agent_toolkits import SQLDatabaseToolkit

toolkit = SQLDatabaseToolkit(db=db, llm=llm)
````

````python
tools = toolkit.get_tools()
tools
````

````python
from langchain_core.messages import SystemMessage

SQL_PREFIX = """You are an agent designed to interact with a DuckDB SQL database.
Given an input question, create a syntactically correct DuckDB SQL query to run, then look at the results of the query and return the answer.
Unless the user specifies a specific number of examples they wish to obtain, always limit your query to at most 5 results.
You can order the results by a relevant column to return the most interesting examples in the database.
Never query for all the columns from a specific table, only ask for the relevant columns given the question.
You have access to tools for interacting with the database.
Only use the below tools. Only use the information returned by the below tools to construct your final answer.
You MUST double check your query before executing it. If you get an error while executing a query, rewrite the query and try again.

DO NOT make any DML statements (INSERT, UPDATE, DELETE, DROP etc.) to the database.

To start you should ALWAYS look at the tables in the database to see what you can query.
Do NOT skip this step.
Then you should query the schema of the most relevant tables."""

system_message = SystemMessage(content=SQL_PREFIX)
````

````python
from langchain_core.messages import HumanMessage
from langgraph.prebuilt import create_react_agent

agent_executor = create_react_agent(llm, tools, state_modifier=system_message, debug=False)
````

````python
question = "How many employees are there?"
# question = "How many departments are there?"

agent_executor.invoke({"messages": [HumanMessage(content=question)]})
````

````python
for s in agent_executor.stream(
    {"messages": [HumanMessage(content=question)]}
):
    print(s)
    print("----")
````
